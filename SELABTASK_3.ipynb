```python
# Weather Prediction using Quadratic Model
# Waterfall Model Representation

import numpy as np
import matplotlib.pyplot as plt

# Function to calculate temperature using quadratic equation
def temp_model(time, p=0.8, q=-1.5, r=28):
    return p * time**2 + q * time + r

# Time range
time_values = np.linspace(-3, 3, 120)

# Predicted temperatures
temp_values = temp_model(time_values)

# Waterfall phases
phases = [
    "Requirement Analysis",
    "System Design",
    "Coding",
    "Testing",
    "Maintenance"
]

# Positions for phases on curve
phase_x = np.linspace(-2.5, 2.5, len(phases))
phase_y = temp_model(phase_x)

# Create plot
plt.figure(figsize=(10, 6))

# Plot quadratic temperature curve
plt.plot(time_values, temp_values, linestyle='--', label="Temperature Curve")

# Connect waterfall phases step-wise
for i in range(1, len(phases)):
    plt.plot([phase_x[i-1], phase_x[i]], [phase_y[i-1], phase_y[i-1]])
    plt.plot([phase_x[i], phase_x[i]], [phase_y[i-1], phase_y[i]])

# Mark phase points
plt.scatter(phase_x, phase_y)

# Label each phase
for i in range(len(phases)):
    plt.text(phase_x[i], phase_y[i] + 0.6, phases[i], ha='center')

# Graph labels
plt.title("Weather Modelling Using Quadratic Equation\n(Waterfall Development Model)")
plt.xlabel("Time / Development Stages")
plt.ylabel("Estimated Temperature")
plt.legend()
plt.grid(True)

# Display graph
plt.show()
```
